```python
import os
import cv2
from IPython.display import Video, display
```

```python
for s in range(1,11):

    input_dir = os.path.expanduser(f"~/Documents/GitHub/Hysteresis_SA/exp_demo/images_31/set{s:02d}")

    # 출력 비디오 파일
    output_path = os.path.expanduser(f"~/Documents/GitHub/Hysteresis_SA/exp_demo/images_31/video/set{s:02d}_video.mp4")

    # frame_01.png ~ frame_31.png
    image_paths = [
        os.path.join(input_dir, f"frame_{i:02d}.png")
        for i in range(1, 32)
    ]

    # 각 프레임 100 ms = 0.1초 → 10 fps
    fps = 10

    # 첫 이미지에서 해상도 확인
    first_frame = cv2.imread(image_paths[0])

    if first_frame is None:
        raise FileNotFoundError(f"이미지를 찾을 수 없습니다: {image_paths[0]}")

    height, width = first_frame.shape[:2]

    # MP4 writer 설정
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    writer = cv2.VideoWriter(
        output_path,
        fourcc,
        fps,
        (width, height)
    )

    # 이미지 순서대로 추가
    for path in image_paths:
        frame = cv2.imread(path)

        if frame is None:
            raise FileNotFoundError(f"이미지를 찾을 수 없습니다: {path}")

        # 혹시 이미지 크기가 다르면 첫 프레임 크기로 맞춤
        if frame.shape[:2] != (height, width):
            frame = cv2.resize(frame, (width, height))

        writer.write(frame)

    writer.release()

    print(f"Saved: {output_path}")
    print(f"Duration: {31 / fps:.2f} seconds")

    # display(Video(output_path, embed=True))
```
